# Precision of the personal email evaluation

This notebook calculates illustrative 95% Wilson intervals for single proportions and normal-approximation half-widths for paired accuracy differences. It is a planning aid, not a power guarantee.

```python
from math import sqrt

Z_95 = 1.959963984540054

def wilson_interval(successes: int, total: int, z: float = Z_95):
    proportion = successes / total
    denominator = 1 + z**2 / total
    centre = (proportion + z**2 / (2 * total)) / denominator
    half_width = z * sqrt(
        proportion * (1 - proportion) / total + z**2 / (4 * total**2)
    ) / denominator
    return centre - half_width, centre + half_width

for total in (100, 200, 300):
    low, high = wilson_interval(round(0.80 * total), total)
    print(total, round(low, 3), round(high, 3), round((high - low) / 2, 3))
```

```python
for total in (100, 200, 300):
    for prevalence in (0.05, 0.10, 0.20):
        count = round(prevalence * total)
        low, high = wilson_interval(count, total)
        print(total, prevalence, count, round(low, 3), round(high, 3))
```

```python
# If models disagree on this share of emails and their observed accuracies are equal,
# the expression gives an illustrative 95% half-width for the paired difference.
for total in (100, 200, 300):
    for discordance_share in (0.10, 0.20, 0.50):
        half_width = Z_95 * sqrt(discordance_share / total)
        print(total, discordance_share, round(half_width, 3))
```
